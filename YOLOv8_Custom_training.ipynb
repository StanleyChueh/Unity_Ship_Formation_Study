```python
import yaml
import os

yaml_path = '/content/datasets/Unity_boat_detection/data.yaml'

with open(yaml_path, 'r') as f:
    data = yaml.safe_load(f)

# Update paths for Colab environment
data['path'] = '/content/datasets/Unity_boat_detection'
data['train'] = 'train/images'
data['val'] = 'val/images'

with open(yaml_path, 'w') as f:
    yaml.dump(data, f)

print("Updated data.yaml paths for Colab.")
```

```python
from ultralytics import YOLO

# Load the pretrained Nano model
model = YOLO('yolov8n.pt')

# Train the model
results = model.train(
    data='/content/datasets/Unity_boat_detection/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='unity_boat_model'
)
```

```python
import shutil

# Copy the best weights to your Drive
source_path = '/content/runs/detect/unity_boat_model/weights/best.pt'
destination_path = '/content/drive/MyDrive/Your_Folder_Path/best_boat_model.pt'

shutil.copy(source_path, destination_path)
print(f"Model saved to Drive at: {destination_path}")
```
